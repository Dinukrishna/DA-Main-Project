```python
# Checking missing values in each column
print("Missing Values Before Handling:")
display(df.isnull().sum())

# Calculate total missing values
total_missing = df.isnull().sum().sum()
print("\nTotal Missing Values:", total_missing)

# Handle missing values only if they exist
if total_missing > 0:

    # Filled numerical columns with their median
    numerical_columns = df.select_dtypes(include=["number"]).columns
    for column in numerical_columns:
        if df[column].isnull().any():
            df[column] = df[column].fillna(df[column].median())
            
    # Filled categorical columns with their mode
    categorical_columns = df.select_dtypes(include=["object", "category"]).columns

    for column in categorical_columns:
        if df[column].isnull().any():
            mode_values = df[column].mode()

            if not mode_values.empty:
                df[column] = df[column].fillna(mode_values[0])

    # Handle missing transaction dates separately
    if df["TransactionDate"].isnull().any():
        df = df.dropna(subset=["TransactionDate"]).copy()

    print("\nMissing values handled.")

else:
    print("\nNo missing values found. No imputation required.")

# Step 4: Verify missing values after handling
print("\nMissing Values After Handling:")
display(df.isnull().sum())

print("\nTotal Missing Values After Handling:",
      df.isnull().sum().sum())
```
